## **Business Background**

**FinGrow Wealth Advisors** is a financial advisory company that manages investment portfolios for thousands of clients across India. Every year, the company reviews each client's investment performance and uses historical data to improve future investment decisions.

The company has collected data from **5,000 investors**. For each investor, information such as age, annual income, investment amount, risk appetite, investment experience, market exposure, portfolio diversification, and previous year's return has been recorded.

The management wants to use Artificial Intelligence (AI) to analyze this data and answer different business questions. Interestingly, the **same dataset** can be used for three different machine learning problems simply by changing the target variable.

---

## Dataset Description

The dataset contains the following variables:

| Variable            | Description                                        |
| ------------------- | -------------------------------------------------- |
| Age                 | Age of the investor (years)                        |
| Income (Lakh ₹)     | Annual income of the investor                      |
| Investment (Lakh ₹) | Total investment amount                            |
| Risk Score          | Risk appetite of the investor (1–10)               |
| Experience          | Investment experience (years)                      |
| Market Exposure (%) | Percentage of investment in equity markets         |
| Diversification     | Number of different asset classes in the portfolio |
| Previous Return (%) | Annual return achieved in the previous year        |

The company has also calculated the actual **Annual Return (%)** earned by each investor.

---

## **Learning Outcome**

This case  demonstrates an important concept in Machine Learning:

* The **input features (X) remain exactly the same**.
* Only the **target variable (y)** changes according to the business requirement.

| Business Requirement                             | Target Variable   | Machine Learning Problem   |
| ------------------------------------------------ | ----------------- | -------------------------- |
| Predict the exact annual return                  | Annual Return (%) | Regression                 |
| Predict whether return is High or Low            | Return Binary     | Binary Classification      |
| Predict whether return is Low, Moderate, or High | Return Category   | Multi-class Classification |

This illustrates that the **type of machine learning problem is determined by the business objective and the target variable—not by the input data.**

## **Business Scenario 1: Regression**
**Business Question**

What annual return (%) is expected for a new investor?

The company wants to estimate the exact annual return that an investor is likely to earn based on the available information.

**Target Variable**

Annual Return (%)

Example

Investor	Predicted Annual Return (%)

a.	18.4

b.	11.7

c.	22.3

Since the target is a continuous numerical value, this is a Regression problem.

## **Business Scenario 2: Binary Classification**

**Business Question**
Instead of predicting the exact return, management now wants a simpler decision.

Will the investor earn a High Return (15% or more) or not?

The annual return is converted into two categories:

- 0 = Low Return (less than 15%)
- 1 = High Return (15% or more)

**Target Variable**

Return Binary

Example

Annual Return (%)	Return Binary
- 18.4	1
- 11.7	0
- 22.3	1

Since there are only two possible classes, this becomes a Binary Classification problem.

## **Business Scenario 3: Multi-class Classification**
**Business Question**

The company now wants to classify investors into three performance categories for portfolio management.

The annual return is divided into:

1. Low Return : Less than 10%
2. Moderate Return : 10% to 20%
3. High Return : More than 20%


These categories are encoded as:

- 0 = Low
- 1 = Moderate
- 2 = High

**Target Variable**

Return Category

Example

Annual Return (%)	Category
- 8.6	Low
- 16.8	Moderate
- 23.5	High

Since there are three classes, this becomes a Multi-class Classification problem.

**Data Set:** https://github.com/YBIFoundation/DeepLearning/raw/main/PredictingInvestmentPortfolioPerformance.csv

## **Solution**

```python
# Step 1: Import Library
import pandas as pd
import numpy as np
```

```python
# Step 2: Read dataset
df = pd.read_csv('https://github.com/YBIFoundation/DeepLearning/raw/main/PredictingInvestmentPortfolioPerformance.csv')
```

```python
df.head()
```

```python
df.info()
```

```python
df.describe()
```

```python
# Step 3: Define y and X
```

```python
df.columns
```

```python
X = df[['age', 'income_lakh', 'investment_lakh', 'risk_score', 'experience',  'market_exposure', 'diversification', 'previous_return']]
```

```python
y_r = df['annual_return'] # no preprocessing
```

```python
y_b = df['return_binary'] # no preprocessing
```

```python
y_m = df['return_category'] # preprocessing OneHotEncoding / to_categorical
```

```python
from keras.utils import to_categorical
```

```python
y_m = to_categorical(y_m)
```

```python
# Step 4: Train test split
from sklearn.model_selection import train_test_split
```

# **Regression**

```python
X_train,X_test,y_train,y_test=train_test_split(X,y_r,random_state=2529)
```

```python
from sklearn.preprocessing import MinMaxScaler
mms = MinMaxScaler()
```

```python
X_train = mms.fit_transform(X_train)
```

```python
X_test = mms.transform(X_test)
```

```python
# Step 5: Build ANN model
from keras.models import Sequential
from keras.layers import Dense
```

```python
model = Sequential()
```

```python
model.add(Dense(8, activation='relu'))
```

```python
model.add(Dense(16, activation='relu'))
```

```python
model.add(Dense(1, activation='linear'))
```

```python
model.compile(optimizer='adam',loss='mean_squared_error')
```

```python
# Step 6: Train model
model.fit(X_train,y_train,epochs=10,batch_size=32)
```

```python
model.summary()
```

```python
# Step 7: Prediction
y_pred = model.predict(X_test)
```

```python
# Step 8: Evaluate
from sklearn.metrics import mean_absolute_error
mean_absolute_error(y_test,y_pred)
```

# **Binary Classification**

```python
X_train,X_test,y_train,y_test=train_test_split(X,y_b,random_state=2529)
```

```python
from sklearn.preprocessing import MinMaxScaler
mms = MinMaxScaler()
```

```python
X_train = mms.fit_transform(X_train)
```

```python
X_test = mms.transform(X_test)
```

```python
# Step 5: Build ANN model
from keras.models import Sequential
from keras.layers import Dense
```

```python
model = Sequential()
```

```python
model.add(Dense(8, activation='relu'))
```

```python
model.add(Dense(16, activation='relu'))
```

```python
model.add(Dense(1, activation='sigmoid'))
```

```python
model.compile(optimizer='adam',loss='binary_crossentropy',metrics=['accuracy'])
```

```python
# Step 6: Train model
model.fit(X_train,y_train,epochs=10,batch_size=32)
```

```python
model.summary()
```

```python
# Step 7: Prediction
y_pred = model.predict(X_test)
```

```python
y_pred = np.round(y_pred)
```

```python
# Step 8: Evaluate
from sklearn.metrics import classification_report
print(classification_report(y_test,y_pred))
```

# **Multi Classification**

```python
X_train,X_test,y_train,y_test=train_test_split(X,y_m,random_state=2529)
```

```python
from sklearn.preprocessing import MinMaxScaler
mms = MinMaxScaler()
```

```python
X_train = mms.fit_transform(X_train)
```

```python
X_test = mms.transform(X_test)
```

```python
# Step 5: Build ANN model
from keras.models import Sequential
from keras.layers import Dense
```

```python
model = Sequential()
```

```python
model.add(Dense(8, activation='relu'))
```

```python
model.add(Dense(16, activation='relu'))
```

```python
model.add(Dense(3, activation='softmax'))
```

```python
model.compile(optimizer='adam',loss='categorical_crossentropy',metrics=['accuracy'])
```

```python
# Step 6: Train model
model.fit(X_train,y_train,epochs=10,batch_size=32)
```

```python
model.summary()
```

```python
# Step 7: Prediction
y_pred = model.predict(X_test)
```

```python
y_pred = np.argmax(y_pred,axis=1)
```

```python
y_test = np.argmax(y_test,axis=1)
```

```python
# Step 8: Evaluate
from sklearn.metrics import classification_report
print(classification_report(y_test,y_pred))
```

# **Results**

The same dataset successfully solves three different machine learning problems.
